```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pickle
import os
```

```python
df = pd.read_csv('../data/train.csv')
```

```python
df.head(10)
```

```python
df.info(show_counts=True)
```

```python
df['blue'] = df['blue'].astype('category')
df['dual_sim'] = df['dual_sim'].astype('category')
df['four_g'] = df['four_g'].astype('category')
df['three_g'] = df['three_g'].astype('category')
df['touch_screen'] = df['touch_screen'].astype('category')
df['wifi'] = df['wifi'].astype('category')
df['price_range'] = df['price_range'].astype('category')
df['fc'] = df['fc'].astype('int8')
df['int_memory'] = df['int_memory'].astype('int8')
df['n_cores'] = df['n_cores'].astype('int8')
df['pc'] = df['pc'].astype('int8')
df['sc_h'] = df['sc_h'].astype('int8')
df['sc_w'] = df['sc_w'].astype('int8')
df['talk_time'] = df['talk_time'].astype('int8')
df['clock_speed'] = df['clock_speed'].astype('float16')
df['m_dep'] = df['m_dep'].astype('float16')
```

```python
df = df.query('px_height >= 20')
df = df.query('sc_w >= 3')
df.describe()
```

```python
df.head(10)
```

```python
# График 1: Зависимость характеристик от ценового диапазона
fig, axs = plt.subplots(2, 2, figsize=(16.5, 14))
fig.tight_layout(pad=3)

sns.lineplot(df, x='price_range', y='battery_power', label='battery_power', ax=axs[0, 0])
axs[0, 0].set_title('Емкость батареи')

sns.lineplot(df, x='price_range', y='int_memory', label='int_memory', ax=axs[0, 1])
axs[0, 1].set_title('Внутренняя память')

sns.lineplot(df, x='price_range', y='ram', label='ram', ax=axs[1, 0])
axs[1, 0].set_title('Оперативная память (RAM)')

sns.lineplot(df, x='price_range', y='fc', label='fc', ax=axs[1, 1])
axs[1, 1].set_title('Фронтальная камера')

plt.savefig('graph1_characteristics_vs_price_range.png', dpi=300, bbox_inches='tight')
print('График сохранен: graph1_characteristics_vs_price_range.png')
plt.show()
```

```python
# График 2: Зависимость размеров экрана от ценового диапазона
fig, axs = plt.subplots(2, 2, figsize=(16.5, 14))
fig.tight_layout(pad=3)

sns.lineplot(df, x='price_range', y='sc_h', label='sc_h', ax=axs[0, 0])
axs[0, 0].set_title('Высота экрана (см)')

sns.lineplot(df, x='price_range', y='sc_w', label='sc_w', ax=axs[0, 1])
axs[0, 1].set_title('Ширина экрана (см)')

sns.lineplot(df, x='price_range', y='px_height', label='px_height', ax=axs[1, 0])
axs[1, 0].set_title('Высота разрешения (px)')

sns.lineplot(df, x='price_range', y='px_width', label='px_width', ax=axs[1, 1])
axs[1, 1].set_title('Ширина разрешения (px)')

plt.savefig('graph2_screen_size_vs_price_range.png', dpi=300, bbox_inches='tight')
print('График сохранен: graph2_screen_size_vs_price_range.png')
plt.show()
```

```python
# Интерактивная визуализация с Bokeh
from bokeh.plotting import figure, show, save
from bokeh.models import ColumnDataSource, HoverTool
```

```python
# График 3: Интерактивная визуализация зависимости цены от ядер и RAM
source = ColumnDataSource(data=dict(
    x=df['n_cores'], 
    y=df['ram'],
    price=df['price_range'], 
    battery=df['battery_power']
))

p = figure(title="Зависимость цены от количества ядер и RAM",
           x_axis_label='Количество ядер',
           y_axis_label='Оперативная память (RAM)',
           tools="pan,wheel_zoom,box_zoom,reset",
           width=800, height=600)

p.scatter('x', 'y', size=10, color="navy", alpha=0.5, source=source)

hover = HoverTool()
hover.tooltips = [("Цена", "@price"), ("RAM", "@y MB"), ("Батарея", "@battery mAh")]
p.add_tools(hover)

save(p, 'graph3_price_vs_cores_ram.html')
print('Интерактивный график сохранен: graph3_price_vs_cores_ram.html')
show(p)
```

```python
# График 4: Зависимость характеристик от толщины устройства
fig, axs = plt.subplots(2, 2, figsize=(16.5, 14))
fig.tight_layout(pad=3)

sns.lineplot(df, x='m_dep', y='mobile_wt', label='mobile_wt', ax=axs[0, 0])
axs[0, 0].set_title('Вес устройства')

sns.lineplot(df, x='m_dep', y='battery_power', label='battery_power', ax=axs[0, 1])
axs[0, 1].set_title('Емкость батареи')

sns.lineplot(df, x='m_dep', y='pc', label='pc', ax=axs[1, 0])
axs[1, 0].set_title('Основная камера')

sns.lineplot(df, x='m_dep', y='n_cores', label='n_cores', ax=axs[1, 1])
axs[1, 1].set_title('Количество ядер')

plt.savefig('graph4_characteristics_vs_thickness.png', dpi=300, bbox_inches='tight')
print('График сохранен: graph4_characteristics_vs_thickness.png')
plt.show()
```

```python
# График 5: Зависимость времени разговора от наличия функций
fig, axs = plt.subplots(2, 2, figsize=(16.5, 14))
fig.tight_layout(pad=3)

sns.lineplot(df, x='wifi', y='talk_time', label='talk_time', ax=axs[0, 0])
axs[0, 0].set_title('WiFi')

sns.lineplot(df, x='blue', y='talk_time', label='talk_time', ax=axs[0, 1])
axs[0, 1].set_title('Bluetooth')

sns.lineplot(df, x='four_g', y='talk_time', label='talk_time', ax=axs[1, 0])
axs[1, 0].set_title('4G')

sns.lineplot(df, x='touch_screen', y='talk_time', label='talk_time', ax=axs[1, 1])
axs[1, 1].set_title('Сенсорный экран')

plt.savefig('graph5_talk_time_vs_features.png', dpi=300, bbox_inches='tight')
print('График сохранен: graph5_talk_time_vs_features.png')
plt.show()
```
